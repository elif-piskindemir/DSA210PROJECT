```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'pressure': np.random.normal(50, 5, 100),
    'flow': np.random.normal(100, 10, 100)
})

df.head()
df.describe()

plt.hist(df['pressure'])
plt.title("Pressure Distribution")
plt.show()

df['anomaly'] = (df['pressure'] < df['pressure'].mean() - 2*df['pressure'].std()).astype(int)
df['anomaly'].value_counts()

normal = df[df['anomaly']==0]['pressure']
abnormal = df[df['anomaly']==1]['pressure']

print("Normal mean:", normal.mean())
print("Anomaly mean:", abnormal.mean())

plt.boxplot([normal, abnormal])
plt.xticks([1,2], ['Normal', 'Anomaly'])
plt.title("Pressure Comparison")
plt.show()
```
